# Exploratory Data Analysis — Network Anomaly Detection

Run `src/preprocess.py` first to create `data/raw/`. This notebook explores the raw data before training.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')

# uses the synthetic file by default; if you're using the real NSL-KDD
# dataset, preprocess.py writes the cleaned version to data/processed/ instead
df = pd.read_csv('../data/raw/network_traffic.csv')
df.head()
```

```python
df.info()
df.describe()
```

## Class balance

```python
plt.figure(figsize=(5,4))
sns.countplot(data=df, x='label', palette='Set2')
plt.title('Normal vs Anomaly count')
plt.show()
```

## Feature distributions by class
`count` (connections in a short window) and `serror_rate` (failed-connection rate) are classic DoS/probe indicators.

```python
fig, axes = plt.subplots(1, 2, figsize=(12,4))
sns.boxplot(data=df, x='label', y='count', ax=axes[0])
axes[0].set_title('Connection count by class')
sns.boxplot(data=df, x='label', y='serror_rate', ax=axes[1])
axes[1].set_title('SYN error rate by class')
plt.tight_layout()
plt.show()
```

## Correlation between numeric features

```python
numeric_cols = ['duration','src_bytes','dst_bytes','count','srv_count','serror_rate','same_srv_rate']
plt.figure(figsize=(7,5))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature correlation heatmap')
plt.show()
```
